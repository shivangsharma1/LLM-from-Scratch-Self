```python
#modified generated function
from torch import topk


def generate(model, idx, max_new_tokens, context_size, temperature = 2, top_k = 3, eos_id = None):
    for _ in range(max_new_tokens):
        idx_cond = idx[:, -context_size:]
        with torch.no_grad():
            logits = model(idx_cond)
        logits = logits[:, -1, :]

        if top_k is not None:
            top_logits, top_logits_idx = torch.topk(logits, top_k)
            min_val = top_logits[:, -1]  # shape: batch, vocab_size logits

            #masking 
            logits = torch.where(
                logits < min_val, 
                torch.tensor(float("-inf")),
                logits
            )

        if temperature > 0.0:
            logits = logits / temperature  #temperature value greater the 1.0, the output will be more uniformly distributed
            probs = torch.softmax(logits, dim = -1)
            idx_next = torch.multinomial(probs, num_samples=1) #num of sample for each batch 
        else:
            idx_next = torch.argmax(logits, dim = -1, keepdim=True)
        
        # if idx_next == eos_id:
        #     break

        idx = torch.cat([idx, idx_next], dim = 1)
    return idx



start_context = "Every effort moves you"
model = GPTModel(contants)
model.eval()
token_ids = generate(
    model=model,
    idx=text_to_token_ids(start_context, tokenizer),
    max_new_tokens=50, 
    context_size=contants.context_length
)

print("output text: \n ", token_ids_to_text(token_ids, tokenizer))
```
